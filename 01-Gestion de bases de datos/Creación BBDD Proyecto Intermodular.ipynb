```python
CREATE DATABASE proyecto intermodular;

USE proyecto intermodular;

CREATE TABLE Alumno (
    id_alumno INT AUTO_INCREMENT PRIMARY KEY,
    nombre VARCHAR(100) NOT NULL,
    email VARCHAR(100) UNIQUE NOT NULL,
    contrasena VARCHAR(255) NOT NULL,
    telefono VARCHAR(20)
);

CREATE TABLE Matricula (
    id_matricula INT AUTO_INCREMENT PRIMARY KEY,
    fecha_matricula DATE,
    estado ENUM('activa','suspendida','finalizada'),
    id_alumno INT,
    FOREIGN KEY (id_alumno) REFERENCES Alumno(id_alumno)
    ON DELETE CASCADE
);

CREATE TABLE Departamento (
    id_departamento INT AUTO_INCREMENT PRIMARY KEY,
    nombre VARCHAR(100),
    equipoHW VARCHAR(100)
);

CREATE TABLE Empleado (
    id_empleado INT AUTO_INCREMENT PRIMARY KEY,
    nombre VARCHAR(100),
    email VARCHAR(100),
    rol VARCHAR(50),
    salario DECIMAL(10,2),
    fecha_contratacion DATE,
    id_departamento INT,
    FOREIGN KEY (id_departamento) REFERENCES Departamento(id_departamento)
);

CREATE TABLE Laboratorio (
    id_laboratorio INT AUTO_INCREMENT PRIMARY KEY,
    nombre VARCHAR(100),
    descripcion TEXT,
    sistema VARCHAR(50),
    dificultad VARCHAR(20),
    id_empleado INT,
    FOREIGN KEY (id_empleado) REFERENCES Empleado(id_empleado)
);

CREATE TABLE Desafio (
    id_desafio INT AUTO_INCREMENT PRIMARY KEY,
    titulo VARCHAR(100),
    descripcion TEXT,
    tipo VARCHAR(50),
    dificultad VARCHAR(20),
    id_laboratorio INT,
    FOREIGN KEY (id_laboratorio) REFERENCES Laboratorio(id_laboratorio)
);

CREATE TABLE Evento_CTF (
    id_evento INT AUTO_INCREMENT PRIMARY KEY,
    nombre VARCHAR(100),
    descripcion TEXT,
    fecha_inicio DATE,
    fecha_fin DATE
);

CREATE TABLE Incidencia (
    id_incidencia INT AUTO_INCREMENT PRIMARY KEY,
    titulo VARCHAR(100),
    descripcion TEXT,
    estado VARCHAR(50),
    fecha_creacion DATE,
    id_alumno INT,
    id_empleado INT,
    FOREIGN KEY (id_alumno) REFERENCES Alumno(id_alumno),
    FOREIGN KEY (id_empleado) REFERENCES Empleado(id_empleado)
);

CREATE TABLE Alumno_Evento (
    id_alumno INT,
    id_evento INT,
    PRIMARY KEY (id_alumno, id_evento),
    FOREIGN KEY (id_alumno) REFERENCES Alumno(id_alumno) ON DELETE CASCADE,
    FOREIGN KEY (id_evento) REFERENCES Evento_CTF(id_evento)
);

CREATE TABLE Alumno_Desafio (
    id_alumno INT,
    id_desafio INT,
    fecha DATE,
    completado BOOLEAN,
    PRIMARY KEY (id_alumno, id_desafio),
    FOREIGN KEY (id_alumno) REFERENCES Alumno(id_alumno) ON DELETE CASCADE,
    FOREIGN KEY (id_desafio) REFERENCES Desafio(id_desafio)
);
```
