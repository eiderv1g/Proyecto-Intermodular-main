```python
SELECT a.nombre, m.estado, m.fecha_matricula
FROM Alumno a
JOIN Matricula m ON a.id_alumno = m.id_alumno;
```

```python
SELECT a.nombre, d.titulo
FROM Alumno a
JOIN Alumno_Desafio ad ON a.id_alumno = ad.id_alumno
JOIN Desafio d ON ad.id_desafio = d.id_desafio
WHERE ad.completado = TRUE;
```

```python
SELECT i.titulo, a.nombre AS alumno, e.nombre AS empleado
FROM Incidencia i
JOIN Alumno a ON i.id_alumno = a.id_alumno
JOIN Empleado e ON i.id_empleado = e.id_empleado;
```
