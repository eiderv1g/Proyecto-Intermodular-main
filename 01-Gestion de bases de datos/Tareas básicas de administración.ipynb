# Explicación de tareas básicas de administración

## Gestión de usuarios

La gestión de usuarios permite controlar quién puede acceder a la base de datos y qué acciones puede realizar.

En MySQL, se pueden crear usuarios y asignarles permisos específicos:

```sql
CREATE USER 'usuario'@'localhost' IDENTIFIED BY 'password';
GRANT ALL PRIVILEGES ON plataforma_ctf.* TO 'usuario'@'localhost';

Con esto onseguimos que se cree un usuario con sus propias credenciales y asignarle todos los privilegios. En caso necesario se podría borrar.


## Copias de seguridad

Las copias de seguridad permiten guardar una copia de la base de datos para evitar la pérdida de información.

En phpMyAdmin se realizan de la siguiente manera:

1. Seleccionar la base de datos  
2. Ir a la pestaña **Exportar**  
3. Elegir formato SQL  
4. Descargar el archivo  

Este archivo puede utilizarse posteriormente para restaurar la base de datos en caso de fallo.

---

## Restauración de la base de datos

Permite recuperar una base de datos a partir de un archivo de copia de seguridad.

En phpMyAdmin:

1. Crear o seleccionar una base de datos  
2. Ir a la pestaña **Importar**  
3. Seleccionar el archivo `.sql`  
4. Ejecutar la importación  

---

## Integridad de datos

Para mantener la coherencia de la información, se utilizan:

- **Claves primarias (PK)**: identifican de forma única cada registro  
- **Claves foráneas (FK)**: establecen relaciones entre tablas  
- Restricciones como `NOT NULL` o `UNIQUE`  

También se pueden usar opciones como:

- `ON DELETE CASCADE` para eliminar automáticamente registros relacionados  
- `RESTRICT` para evitar borrados si existen dependencias  

---

## Conclusión

Estas tareas básicas de administración permiten garantizar el correcto funcionamiento de la base de datos, asegurando la integridad de los datos, el control de accesos y la disponibilidad de la información en caso de fallo.
