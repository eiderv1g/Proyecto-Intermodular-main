# Análisis de datos

En este proyecto se ha realizado un análisis de los datos necesarios para diseñar una base de datos orientada a una plataforma de formación en ciberseguridad.

El objetivo es crear un sistema que permita gestionar tanto el aprendizaje online como la organización interna de la empresa.

La plataforma está inspirada en entornos reales de aprendizaje práctico, donde los usuarios pueden interactuar con laboratorios, resolver desafíos y participar en eventos.

## Funcionalidades del sistema

El sistema permite lo siguiente:

- Gestionar alumnos y su información personal  
- Controlar las matrículas y el acceso a la formación  
- Administrar empleados y departamentos internos  
- Gestionar laboratorios y desafíos prácticos  
- Permitir la participación en eventos tipo CTF  
- Registrar incidencias y soporte técnico  

## Entidades principales

Tras el análisis, se han identificado las siguientes entidades principales:

- **Alumno**: representa a los usuarios que utilizan la plataforma  
- **Matrícula**: controla el estado del alumno dentro del sistema  
- **Empleado**: personal que gestiona la plataforma  
- **Departamento**: organización interna de la empresa  
- **Laboratorio**: entorno práctico de aprendizaje  
- **Desafío**: actividades dentro de los laboratorios  
- **Evento_CTF**: competiciones de ciberseguridad  
- **Incidencia**: sistema de soporte técnico  

## Consideraciones adicionales

Se ha tenido en cuenta la posibilidad de ampliar el sistema con nuevas funcionalidades como equipos o roles avanzados. 

Además, se ha considerado el uso de restricciones como `ON DELETE CASCADE` para mantener la integridad referencial, aunque finalmente no se ha aplicado para evitar la eliminación accidental de datos.

Por último, el diseño permite realizar consultas complejas mediante el uso de `JOIN`, entre otras operaciones.

## Conclusión

El diseño realizado permite una implementación eficiente en sistemas gestores de bases de datos relacionales como MySQL, facilitando tanto la gestión de datos como la realización de consultas complejas.
